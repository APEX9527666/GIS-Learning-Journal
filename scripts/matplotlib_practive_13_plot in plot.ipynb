```python
import matplotlib.pyplot as plt

fig = plt.figure()
x = [1, 2, 3, 4, 5, 6, 7]
y = [1, 3, 4, 2, 5, 8, 6]

# below are all percentage
left, bottom, width, height = 0.1, 0.1, 0.8, 0.8 # 坐标轴左侧距离画布左边缘 10%，坐标轴底部距离画布下边缘 10%，坐标轴宽度占画布的 80%，坐标轴高度占画布的 80%
ax1 = fig.add_axes([left, bottom, width, height])  # 在画布上添加一个主坐标轴ax1
ax1.plot(x, y, 'r')
ax1.set_xlabel('x')
ax1.set_ylabel('y')
ax1.set_title('title')

ax2 = fig.add_axes([0.2, 0.6, 0.25, 0.25])  # 添加小坐标轴ax2
ax2.plot(y, x, 'b')
ax2.set_xlabel('x')
ax2.set_ylabel('y')
ax2.set_title('title inside 1')


# 添加坐标轴的不同方法
####################################
plt.axes([0.6, 0.2, 0.25, 0.25]) # 左边界0.6，下边界0.2，宽0.25，高0.25
plt.plot(y[::-1], x, 'g') # y用的是列表切片语法，表示将y反转，绘制反转后的 y 和 x 的折线图
plt.xlabel('x')
plt.ylabel('y')
plt.title('title inside 2')

plt.show()
```
