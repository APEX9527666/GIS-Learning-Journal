```python
import matplotlib.pyplot as plt
import numpy as np

n = 1024    # 设置数据点数量
X = np.random.normal(0, 1, n) # 生成均值为0，方差为1的随机数作为散点的x,y坐标
Y = np.random.normal(0, 1, n)
T = np.arctan2(Y, X)    # 计算每个点相对于原点的极角（弧度制），结果范围在[-pi,pi]之间，这个值用于散点的颜色参数

plt.figure('special')
plt.scatter(X, Y, s=75, c=T, alpha=0.5) #s控制散点大小,C=T（用T的值给散点自动着色）,透明度0.5
plt.figure('normal')
plt.scatter(np.arange(2),np.arange(2)) 

plt.xlim(-1.5, 1.5) # 设置x的显示范围
plt.xticks(())  # 清空x轴刻度
plt.ylim(-1.5, 1.5) # 设置y的显示范围
plt.yticks(())  # 清空y轴刻度

plt.show()
```
