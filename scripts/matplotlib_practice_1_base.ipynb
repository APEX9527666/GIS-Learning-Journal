```python
#导入模板重命名
import matplotlib.pyplot as plt
import numpy as np
x=np.linspace(-1,1,50) #将1到-1平分成50段，每段1个点
y=2*x+1
plt.plot(x,y) #绘图
plt.show() #显示图像
```

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.linspace(-3,3,50)
y1=2*x+1
y2=x**2

##figure有顺序

plt.figure(num=1) #图像1 
plt.plot(x,y1)

plt.figure(num=3,figsize=(8,5)) #图像2 名称3 长8宽5
plt.plot(x,y2) 
plt.plot(x,y1,color='red',linewidth=2.0,linestyle='--') #一个图可以有多条线 设置颜色为红色（默认蓝色） 线宽1.0 虚线

plt.show()
```
