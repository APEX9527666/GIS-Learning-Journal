```python
import matplotlib.pyplot as plt
import numpy as np

# 图像数据
a = np.array([0.313660827978, 0.365348418405, 0.423733120134, # np.array 创建一个包含九个浮点数的一堆数组
              0.365348418405, 0.439599930621, 0.525083754405,
              0.423733120134, 0.525083754405, 0.651536351379]).reshape(3,3) # .reshape(3,3)把这个一维数组重塑为三行三列的二维矩阵

"""
关于 "interpolation"（插值方式）取值:
http://matplotlib.org/examples/images_contours_and_fields/interpolation_methods.html
关于"origin"= ['upper', 'lower']（原点）取值, 查看:
http://matplotlib.org/examples/pylab_examples/image_origin.html
"""
plt.imshow(a, interpolation='nearest', cmap='bone', origin='lower') #.imshow显示二维数组为图像的函数，插值方式为最接近邻插值
# 即（让每个数据点显示为清晰的方块，不做平滑处理），颜色映射为bone，origin设置图像的坐标原点，lower表示把矩阵的第一行放在图像的底部
plt.colorbar(shrink=.92) # colorbar在图像旁添加一个颜色条，用来显示颜色和数值的对应关系，shrink把颜色条长度缩小为原图的92%

plt.xticks(())
plt.yticks(())
plt.show()
```
