```python
import numpy as np
import matplotlib.pyplot as plt


fig = plt.figure()
ax = fig.add_subplot(111, projection='3d') # add_subplot 是画布自带的标准子图创建函数，111把画布分成 1 行 1 列，使用第 1 个子图，
# projection指定这个子图的投影模式为 3D 立体投影
# X, Y value
X = np.arange(-4, 4, 0.25)
Y = np.arange(-4, 4, 0.25)
X, Y = np.meshgrid(X, Y) # 把一堆x,y的数组转换成二维网格矩阵，以方便计算每个网格点到原点的距离
R = np.sqrt(X ** 2 + Y ** 2) # 计算每个网格点到原点的距离（半径）

Z = np.sin(R) # 用.sin(R)生成高度值，得到3D曲面数据

ax.plot_surface(X, Y, Z, rstride=1, cstride=1, cmap=plt.get_cmap('rainbow')) #（x,y,z）是以二维数组形式的坐标和高度数据，设置rstride（行步长）为1
# 即每隔一行绘制一条曲面线（数值越小线越密），cstridde（列步长）为1,表示每隔一列绘制一条曲面线（数值越小曲面越密），cmap使用彩虹颜色映射，让曲面颜色随z值变化
"""
============= ================================================
        参数           描述
        ============= ================================================
        *X*, *Y*, *Z* 以二维数组形式存在的数据值
        *rstride*     数组行步长（步长大小），默认值为 10
        *cstride*     数组列步长（步长大小），默认值为 10
        *color*       曲面补丁的颜色
        *cmap*        曲面补丁使用的颜色映射.
        *facecolors*  每个独立补丁的面颜色
        *norm*        用于将数值映射到颜色的归一化实例
        *vmin*        颜色映射的最小值
        *vmax*        颜色映射的最大值
        *shade*       是否对面颜色进行光影着色
        ============= ================================================
"""

# 设置底面等高线图（注意与之前等高线不太一样）
ax.contourf(X, Y, Z, zdir='z', offset=-2, cmap=plt.get_cmap('rainbow'))
"""
==========  ================================================
        参数         描述
        ==========  ================================================
        *X*, *Y*,   以 numpy 数组形式存储的数据值
        *Z*
        *zdir*      投影方向：x、y 或 z（默认是 z）
        *offset*    如果指定该参数，则在垂直于zdir的平面上的这个位置绘制填充等高线的投影
        ==========  ================================================
"""

ax.set_zlim(-2, 2)

plt.show()
```
