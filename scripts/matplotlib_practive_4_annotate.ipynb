```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-3, 3, 50)
y = 2*x + 1

plt.figure(num=1, figsize=(8, 5),)
plt.plot(x, y,) #.scatter为散点图

ax = plt.gca()
ax.spines['right'].set_color('none')
ax.spines['top'].set_color('none')
ax.spines['top'].set_color('none')
ax.xaxis.set_ticks_position('bottom')
ax.spines['bottom'].set_position(('data', 0))
ax.yaxis.set_ticks_position('left')
ax.spines['left'].set_position(('data', 0))

x0 = 1
y0 = 2*x0 + 1
plt.plot([x0, x0,], [0, y0,], 'k--', linewidth=2.5) # x从x0到x0，y从0到y0,k为‘black’的简写,'--'表虚线
plt.scatter([x0, ], [y0, ], s=50, color='b')

# 怎么用annotation?

# 方式 1: annotate用于打箭头
#####################
plt.annotate(r'$2x+1=%s$' % y0, xy=(x0, y0), xycoords='data', xytext=(+30, -30), # xy=(x0,y0)表示1箭头指向的目标坐标点,data使用图标本身数据坐标系,text表偏移量
             textcoords='offset points', fontsize=16, # textcoords 设置偏移单位为像素点，而不是x/y轴数据刻度，fontsize设置标注字号为16号
             arrowprops=dict(arrowstyle='->', connectionstyle="arc3,rad=.2")) # arrowstyle设置箭头形状（<-,<->,-|>等样式），connectionstyle控制线的弯曲程度

# 方式 2: text用于在指定坐标写文字
########################
plt.text(-3.7, 3, r'$This\ is\ the\ some\ text. \mu\ \sigma_i\ \alpha_t$', # 用text定义在x=-3.7,y=3处放置文字
         
         fontdict={'size': 16, 'color': 'r'})

plt.show()
```
