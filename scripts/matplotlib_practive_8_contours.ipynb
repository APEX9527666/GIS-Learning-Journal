```python
import matplotlib.pyplot as plt
import numpy as np

def f(x,y):
    # 输入xy两个点计算高度
    return (1 - x / 2 + x**5 + y**3) * np.exp(-x**2 -y**2)

n = 256
x = np.linspace(-3, 3, n) # 在[-3,3]区间生成256个等间距的数值
y = np.linspace(-3, 3, n)
X,Y = np.meshgrid(x, y) # 将一组数组x和y转换成二维网格矩阵x和y，（等高线图底图还是网格），用于后续计算每个网格点的高度

# 用 plt.contourf 填充轮廓
# X, Y and value for (X,Y) point
plt.contourf(X, Y, f(X, Y), 8, alpha=.75, cmap=plt.cm.hot) # 8代表将高度范围分成8个等级，对应8种颜色填充，alpha设置颜色透明度，hot颜色映射（从黑到红到黄的暖色调）

# 用 plt.contour 添加等高线
C = plt.contour(X, Y, f(X, Y), 8, colors='black', linewidths=.5) # 8代表生成8条等高线，等高线颜色为黑色，线宽为0.5
# 添加数值标注
plt.clabel(C, inline=True, fontsize=10) #标注文字嵌入到等高线内部，标注文字大小为10 (False则让数值穿过数字)

plt.xticks(())
plt.yticks(())
plt.show()
```
