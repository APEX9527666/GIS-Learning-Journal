matplotlib设置坐标轴
时间：2026-6-24
练习内容：设置坐标轴取值范围、特殊值、艺术字体，图像边框

```python
import numpy as np
import matplotlib.pyplot as plt
x=np.linspace(-3,3,50)
y1=2*x+1
y2=x**2

plt.figure()
plt.plot(x,y2)
plt.plot(x,y1,color='red',linewidth=3.0,linestyle='--')

plt.xlim((-1,2)) #设置x取值范围
plt.ylim((-2,3)) #设置y取值范围
plt.xlabel('I am x') #描述x轴
plt.ylabel('I am y') #描述y轴

new_ticks=np.linspace(-1,2,5) 
print(new_ticks)
plt.xticks(new_ticks) #设置x的ticks(小标)
plt.yticks([-2,-1.8,-1,1.22,3,],[r'$really\ bad$',r'$bad$',r'$normal\ \alpha$',r'$good$',r'$really\ good$']) #修改yticks的不同对应值,字母前后加$可将字体修改成艺术字体
#'\ '可在子母间插入一个空格，r用于让字符串中的转义字符失效，即让Python将'\ '当作正常空格，\alpha可输出数学符号alpha、--

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
x=np.linspace(-3,3,50)
y1=2*x+1
y2=x**2

plt.figure()
plt.plot(x,y2)
plt.plot(x,y1,color='red',linewidth=3.0,linestyle='--')

plt.xlim((-1,2)) #设置x取值范围
plt.ylim((-2,3)) #设置y取值范围
plt.xlabel('I am x') #描述x轴
plt.ylabel('I am y') #描述y轴

new_ticks=np.linspace(-1,2,5) 
print(new_ticks)
plt.xticks(new_ticks) #设置x的ticks(小标)
plt.yticks([-2,-1.8,-1,1.22,3,],[r'$really\ bad$',r'$bad$',r'$normal\ \alpha$',r'$good$',r'$really\ good$']) #修改yticks的不同对应值,字母前后加$可将字体修改成艺术字体
#'\ '可在子母间插入一个空格，r用于让字符串中的转义字符失效，即让Python将'\ '当作正常空格，\alpha可输出数学符号alpha、--

# gca='get current axis'(获取当前坐标)
ax=plt.gca()
ax.spines['right'].set_color('none') #(脊梁)修改 将右边框设置为‘没有’
ax.spines['top'].set_color('none') #上边框设置为‘没有’
ax.xaxis.set_ticks_position('bottom') #用下坐标轴代替x的小标
ax.yaxis.set_ticks_position('left') #用左坐标轴代替y的小标
ax.spines['bottom'].set_position(('data',0)) #设置下坐标轴的位置 0为坐标轴的原点，x周绑定在y轴的0位置
ax.spines['left'].set_position(('data',0)) #设置0为坐标轴的位置

#定位 outward(文本、标签朝外),axes(定位到百分之多少)

plt.show()
```
