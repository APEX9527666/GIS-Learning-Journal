```python
import matplotlib.pyplot as plt
import numpy as np

n = 12
X = np.arange(n) # 生成一个从0到11的数组，作为柱子的x轴位置
# 生成两组数值范围在(0.5/12,1.0)之间的随机数据，整体趋势都是从左到右递减
Y1 = (1 - X / float(n)) * np.random.uniform(0.5, 1.0, n)
Y2 = (1 - X / float(n)) * np.random.uniform(0.5, 1.0, n)

plt.bar(X, +Y1, facecolor='#9999ff', edgecolor='white') #绘制柱状图,第一个参数x表柱子的x轴位置，第二个参数+y表柱子高度（+向上-向下），fa.指填充颜色.ed.指边框颜色
plt.bar(X, -Y2, facecolor='#ff9999', edgecolor='white')

for x, y in zip(X, Y1): # 把x和y1中对应位置的颜色配对，遍历每个侄子的位置和高度
    # ha: 水平对齐
    # va: 垂直对齐
    plt.text(x + 0.4, y + 0.05, '%.2f' % y, ha='center', va='bottom') # xy调整位置，水平对齐方式为中心对齐，垂直对齐方式为底部对齐

for x, y in zip(X, Y2):
    # ha: horizontal alignment
    # va: vertical alignment
    plt.text(x + 0.4, -y - 0.05, '%.2f' % y, ha='center', va='top')

plt.xlim(-.5, n)
plt.xticks(()) #隐藏刻度和标签
plt.ylim(-1.25, 1.25)
plt.yticks(())

plt.show()
```
