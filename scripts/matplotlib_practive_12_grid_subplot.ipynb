```python
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec

# method 1: subplot2grid（基于网格的布局方式）
# plt.subplot2grid(shape, loc, rowspan=1, colspan=1) -> （网格的行列数，子图起始位置、子图占据的行数 / 列数）
##########################
plt.figure()
ax1 = plt.subplot2grid((3, 3), (0, 0), colspan=3)  # 3行3列，行、列索引从 0 开始，占3行

ax1.plot([1, 2], [1, 2])
ax1.set_title('ax1_title')
ax2 = plt.subplot2grid((3, 3), (1, 0), colspan=2)
ax3 = plt.subplot2grid((3, 3), (1, 2), rowspan=2)
ax4 = plt.subplot2grid((3, 3), (2, 0))
ax4.scatter([1, 2], [2, 2])
ax4.set_xlabel('ax4_x')
ax4.set_ylabel('ax4_y')
ax5 = plt.subplot2grid((3, 3), (2, 1))

# method 2: gridspec（跨行跨列写法）
#########################
plt.figure()
gs = gridspec.GridSpec(3, 3) # 实例化网格对象 gs，将当前画布划分为 3 行 × 3 列 的网格；后续子图直接切片 gs[行切片, 列切片] 控制占格范围。
# 网格行、列索引全部从0开始计数
ax6 = plt.subplot(gs[0, :]) # 第 0 行，所有列（: 代表全部），等价于 colspan=3；用 plt.subplot() 绑定网格切片生成子图，赋值 ax6
ax7 = plt.subplot(gs[1, :2]) # 第 1 行，列取 0、1（:2 代表到第 2 列之前，不包含 2），横向占 2 列
ax8 = plt.subplot(gs[1:, 2]) # 行从 1 到最后一行（1、2 行），固定第 2 列，纵向占 2 行
ax9 = plt.subplot(gs[-1, 0]) # -1 代表最后一行（第 2 行），第 0 列
ax10 = plt.subplot(gs[-1, -2]) # 最后一行，倒数第二列（第 1 列）

# method 3: easy to define structure（快速均匀分图（适合行列均等、无跨行跨列））
####################################
f, ((ax11, ax12), (ax13, ax14)) = plt.subplots(2, 2, sharex=True, sharey=True)
# plt.subplots(2,2)：直接创建一张画布 + 2 行 2 列 共 4 个均等大小子图；返回两个值：画布对象 f、子图二维数组 ((第一行两个ax),(第二行两个ax))；
# sharex=True, sharey=True：4 张子图共享同一套 X 轴、Y 轴刻度，减少冗余坐标轴。解构赋值：f 接收画布，ax11/ax12 第一行左右，ax13/ax14 第二行左右
ax11.scatter([1,2], [1,2])

plt.tight_layout()
plt.show()
```
