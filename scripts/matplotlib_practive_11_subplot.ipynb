```python
import matplotlib.pyplot as plt

# 例 1:
###############################                                                   # 子图排列顺序（从上到下。从左到右） 
plt.figure(figsize=(6, 4)) # 创建一张宽6英寸，高4英寸的空白画布                      #        1    |     2
# plt.subplot(_行, _列, 第_个子图)                                                 # -------------------------------
plt.subplot(2, 2, 1) # 将画布分为2行2列共4个子图，选中第1个子图                      #        3    |     4
plt.plot([0, 1], [0, 1]) # 画一条x在[0,1],y在[0.1]的直线

plt.subplot(222) # 2行2列2个子图
plt.plot([0, 1], [0, 2]) # x从0->1,y从0->2

plt.subplot(223) 
plt.plot([0, 1], [0, 3]) 

plt.subplot(224)
plt.plot([0, 1], [0, 4])

plt.tight_layout() # 自动调整所有子图的间距，避免坐标轴标题、图框重叠，美化排版

# 例 2:
###############################
plt.figure(figsize=(6, 4))
# plt.subplot(_行, _列, 第_个子图)  
plt.subplot(2, 1, 1)
# 画布划分为2行1列，总共2个子图，选中第1个（上半部分大图）
plt.plot([0, 1], [0, 1])

plt.subplot(234)
# 重新把画布切分为2行3列，共6个子图，选中第4张子图
plt.plot([0, 1], [0, 2])

plt.subplot(235)
# 同上，选中5张子图
plt.plot([0, 1], [0, 3])

plt.subplot(236)
# 同上，选中第6张子图
plt.plot([0, 1], [0, 4])


plt.tight_layout()
plt.show()
```
