```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(0, 10, 0.1)
y1 = 0.05 * x**2
y2 = -1 *y1

fig, ax1 = plt.subplots() # 创建画布和主坐标轴 ax1

ax2 = ax1.twinx()    # 创建一个共享 x 轴的次坐标轴 ax2，两者 x 轴范围相同，但 y 轴相互独立

ax1.plot(x, y1, 'g-') # g-表绿实线
ax2.plot(x, y2, 'b-')

ax1.set_xlabel('X data')
ax1.set_ylabel('Y1 data', color='g')
ax2.set_ylabel('Y2 data', color='b')

plt.show()
```
