```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-3, 3, 50)
y1 = 2*x + 1
y2 = x**2

plt.figure()
plt.xlim((-1, 2))
plt.ylim((-2, 3))

new_sticks = np.linspace(-1, 2, 5)
plt.xticks(new_sticks)

plt.yticks([-2, -1.8, -1, 1.22, 3],
           [r'$really\ bad$', r'$bad$', r'$normal$', r'$good$', r'$really\ good$'])

l1, = plt.plot(x, y1, label='linear line') #label用于对线命名
l2, = plt.plot(x, y2, color='red', linewidth=1.0, linestyle='--', label='square line')

plt.legend(loc='upper right') #.legend打印图例 (常用参数handles（绘图元素对象）,labels,loc（控制图例在图中的位置）)
# plt.legend(handles=[l1, l2], labels=['up', 'down'],  loc='best') handels[1,2]与labels[a,b]一一对应
#  "," 在 l1, = plt... and l2, = plt... 非常重要 
"""
    loc 参数的位置代码对照表:

          'best' : 0,          (currently not supported for figure legends)
          'upper right'  : 1,
          'upper left'   : 2,
          'lower left'   : 3,
          'lower right'  : 4,
          'right'        : 5,
          'center left'  : 6,
          'center right' : 7,
          'lower center' : 8,
          'upper center' : 9,
          'center'       : 10,"""

plt.show()
```
